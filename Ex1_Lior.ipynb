#### 8. 
##### Write two functions, the first function takes an integer and returns (but does not print) its factorial, the second function takes an integer and prints it to the screen with the text: "The answer is".

##### Below the functions, write a program where the user is asked to provide an integer N in the range of 1-10. The program passes this number to the function that calculates N! and returns the result. The result is then passed to the second function, which prints it to the screen. If the number is outside the allowed range, the program should notify the user and not print anything.

```python
def calc_factorial(n):
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    return fact

def print_result(result):
    print("The answer is", result)

n = int(input("Choose a number 1-10: "))
if 1 <= n <= 10:
    ans = calc_factorial(n)
    print_result(ans)
else:
    print("You are out of range")
```

#### 9. 
##### Write a program that receives exactly 10 positive integers from the user, one after another. The program stores the numbers in a list, and once the input is complete, it prints the following:

   - The list of numbers.
   - The list in reverse order.
   - The list sorted in ascending order.
   - The list sorted in descending order.
   - The sum of the list.
   - The average of the list.
   - A list of unique numbers (with duplicates removed), sorted in ascending order.

##### Add a text field after the outputs where you explain how the solution for the last item works.

```python
numbers = []

for i in range(10):
    n = int(input("Enter a positive integer: "))
    numbers.append(n)

print("1. The list of numbers:", numbers)
print("2. The list in reverse order:", numbers[::-1])
print("3. The list sorted in ascending order:", sorted(numbers))
print("4. The list sorted in descending order:", sorted(numbers, reverse=True))
print("5. The sum of the list:", sum(numbers))
print("6. The average of the list:", sum(numbers) / len(numbers))

unique_list = sorted(list(set(numbers)))
print("7. A list of unique numbers (with duplicates removed), sorted in ascending order:", unique_list)
```

The solution for the unique list works by first converting the list into a set using `set(numbers)`. A set inherently cannot contain duplicate elements, so any duplicates are immediately removed. Then, `list()` converts the set back into a list format, and `sorted()` arranges the remaining unique numbers in ascending order from smallest to largest.

##### iv. Write a function `find_score(student_name)` that:
   
   - Takes a student's name as input.
   - Returns their score if they are in the dictionary.
   - If the student is not found, return the message: `"Student not found."`

    test the function on an existing student name and on a non-existing student name

##### v. Write a loop that prints out each student's name and their score in the format:

       Alice: 85
       Bob: 95

##### vi. Delete Charlie from the student_scores dictionary.

```python
student_scores = {'Alice': 85, 'Bob': 90, 'Charlie': 78, 'Diana': 92}

student_scores['Eve'] = 88
student_scores['Bob'] = 95

def find_score(student_name):
    if student_name in student_scores:
        return student_scores[student_name]
    else:
        return "Student not found."

print(find_score("Alice"))
print(find_score("David"))

for name, score in student_scores.items():
    print(f"{name}: {score}")

del student_scores['Charlie']
```
